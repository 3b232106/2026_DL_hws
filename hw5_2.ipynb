# f. 建立predict_review函數

```python
def predict_review(input_text):
  input_seq = tokens.texts_to_sequences([input_text])
  pad_input_seq  = sequence.pad_sequences(input_seq , maxlen=100)
  predict_result=model.predict(pad_input_seq)
  SentimentDict={1:'正面的',0:'負面的'}
  result = SentimentDict[round(predict_result[0][0])]
  return result
```

# h. 用 Gradio 打造情意分析 Web App

```python
import gradio as gr

def predict_review(input_text):
    input_seq = tokens.texts_to_sequences([input_text])
    pad_input_seq = sequence.pad_sequences(input_seq, maxlen=100)
    predict_result = model.predict(pad_input_seq)

    SentimentDict = {1:'正面的', 0:'負面的'}
    result = SentimentDict[round(predict_result[0][0])]
    return result

demo = gr.Interface(
    fn=predict_review,
    inputs=gr.Textbox(label="輸入評論", placeholder="請輸入一段評論..."),
    outputs=gr.Textbox(label="情緒分析結果"),
    title="評論情緒分析",
    description="輸入一段評論，我會幫你判斷評論是正面還是負面。"
)

demo.launch()
```
